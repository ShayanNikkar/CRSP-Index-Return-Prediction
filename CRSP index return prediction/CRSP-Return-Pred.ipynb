```python
selected_columns = ['dfy', 'infl', 'svar', 'de', 'lty', 'tms', 'tbl', 
                    'dfr', 'dp', 'dy', 'ltr', 'ep', 'b/m', 'ntis', 'mr']

df_selected = df[selected_columns]

# Initialize an empty DataFrame for the new features
df_new_features = pd.DataFrame(index=df.index)

# 1. Create lag features (1-month, 2-month, 3-month lags)
for col in selected_columns:
    for lag in range(1, 4):
        df_new_features[f'{col}_lag_{lag}'] = df_selected[col].shift(lag)

# 2. Create rolling statistics (rolling mean, std, min, max over 3, 6, 12 months)
windows = [3, 6, 12]
for col in selected_columns:
    for window in windows:
        df_new_features[f'{col}_rolling_mean_{window}'] = df_selected[col].rolling(window=window, min_periods=1).mean()
        df_new_features[f'{col}_rolling_std_{window}'] = df_selected[col].rolling(window=window, min_periods=1).std()
        df_new_features[f'{col}_rolling_min_{window}'] = df_selected[col].rolling(window=window, min_periods=1).min()
        df_new_features[f'{col}_rolling_max_{window}'] = df_selected[col].rolling(window=window, min_periods=1).max()

# 3. Create interaction features (pairwise sums, differences, products, ratios)
from itertools import combinations

for col1, col2 in combinations(selected_columns, 2):
    df_new_features[f'{col1}_plus_{col2}'] = df_selected[col1] + df_selected[col2]
    df_new_features[f'{col1}_minus_{col2}'] = df_selected[col1] - df_selected[col2]
    df_new_features[f'{col1}_times_{col2}'] = df_selected[col1] * df_selected[col2]
    df_new_features[f'{col1}_div_{col2}'] = df_selected[col1] / (df_selected[col2] + 1e-6)  # Avoid division by zero

# 4. Create polynomial features (squared and cubic terms)
for col in selected_columns:
    df_new_features[f'{col}_squared'] = df_selected[col] ** 2
    df_new_features[f'{col}_cubed'] = df_selected[col] ** 3

df_new_features
```

```python
# Add the new features to the original DataFrame (excluding 'R' and 'date')
df_combined = pd.concat([df, df_new_features], axis=1)

# Display the shape and the first few rows of the combined DataFrame
df_combined.shape, df_combined.head()

# Move the 'R' column to the last position
R_column = df_combined.pop('R')
df_combined['R'] = R_column

df_combined
```

```python
df_combined.fillna(0, inplace=True)
```

```python
df_rolling_zscore = df_combined.copy()
excluded_columns = ['date', 'R']
columns_to_normalize = [col for col in df_combined.columns if col not in excluded_columns]

# Rolling z-score function
def rolling_zscore(series, window=50):
    return (series - series.rolling(window=window, min_periods=1).mean()) / series.rolling(window=window, min_periods=1).std()

# Apply rolling z-score and clip values to the range [-4, 4]
for col in columns_to_normalize:
    df_rolling_zscore[col] = rolling_zscore(df_rolling_zscore[col], window=50)
    df_rolling_zscore[col] = df_rolling_zscore[col].clip(-4, 4)

# Drop the first 60 rows and retain 'date' and 'R'
df_final = df_rolling_zscore.drop(index=df_rolling_zscore.index[:60])

df_final
```

```python
df_final['date'] = pd.to_datetime(df_final['date'], errors='coerce')

# Split the data into train and test sets based on the date range
train_df = df_final[df_final['date'].dt.year <= 2018]
test_df = df_final[df_final['date'].dt.year >= 2019]
```

```python
from sklearn.linear_model import Ridge
from sklearn.metrics import r2_score
import optuna

# Define the train and test split function based on rolling expanding window
def rolling_expanding_ridge(df, alphas, initial_train_years=4, test_years=4):
    results = []
    predictions = []
    actuals = []

    # Ensure 'date' is in datetime format
    df['date'] = pd.to_datetime(df['date'])

    # Sort by date
    df.sort_values(by='date', inplace=True)

    # Start year for the rolling expanding window
    start_year = df['date'].dt.year.min()

    # End year for the rolling expanding window
    end_year = df['date'].dt.year.max()

    # Loop through the years for rolling expanding window
    while start_year + initial_train_years + test_years <= end_year:
        train_mask = (df['date'].dt.year >= start_year) & (df['date'].dt.year < start_year + initial_train_years)
        test_mask = (df['date'].dt.year >= start_year + initial_train_years) & (df['date'].dt.year < start_year + initial_train_years + test_years)

        train_data = df[train_mask].drop(columns=['date'])
        test_data = df[test_mask].drop(columns=['date'])

        X_train = train_data.drop(columns='R')
        y_train = train_data['R']
        X_test = test_data.drop(columns='R')
        y_test = test_data['R']

        # Train Ridge model with the given alpha
        best_r2 = -np.inf
        best_alpha = None
        for alpha in alphas:
            model = Ridge(alpha=alpha)
            model.fit(X_train, y_train)
            y_pred = model.predict(X_test)
            r2 = r2_score(y_test, y_pred)
            if r2 > best_r2:
                best_r2 = r2
                best_alpha = alpha

        # Train final model with the best alpha
        model = Ridge(alpha=best_alpha)
        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)

        # Store the results
        predictions.extend(y_pred)
        actuals.extend(y_test)

        # Update the start year
        start_year += test_years

    return predictions, actuals

# Define the objective function for Optuna
def objective(trial):
    alphas = [trial.suggest_int('alpha', 1, 10000)]
    predictions, actuals = rolling_expanding_ridge(train_df.copy(), alphas)
    return r2_score(actuals, predictions)

# Run Optuna to find the best alpha
study = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=100)

# Get the best alpha
best_alpha = study.best_params['alpha']
best_alpha
```

```python
# Step 1: Train Ridge Regression on the full train_df and get the coefficients
ridge_model_full = Ridge(alpha=10000)
X_train_full = train_df.drop(columns=['date', 'R'])
y_train_full = train_df['R']

ridge_model_full.fit(X_train_full, y_train_full)

# Get the coefficients and feature names
coefficients = ridge_model_full.coef_
feature_names = X_train_full.columns

# Create a DataFrame of feature names and their corresponding coefficients
coeff_df = pd.DataFrame({
    'Feature': feature_names,
    'Coefficient': coefficients
})

# Sort by the absolute value of the coefficients
coeff_df['abs_coefficient'] = coeff_df['Coefficient'].abs()
top_features = coeff_df.sort_values(by='abs_coefficient', ascending=False).head(200)['Feature'].tolist()

# Plot the top 200 coefficients
plt.figure(figsize=(12, 8))
top_coeff_df = coeff_df[coeff_df['Feature'].isin(top_features)].sort_values(by='Coefficient', ascending=False)
plt.barh(top_coeff_df['Feature'], top_coeff_df['Coefficient'], color='skyblue')
plt.xlabel('Coefficient Value')
plt.title('Top 200 Features by Ridge Regression Coefficients')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()

# Step 2: Filter train_df and test_df to retain only top 200 features + 'date' and 'R'
columns_to_keep = ['date', 'R'] + top_features
train_df_filtered = train_df[columns_to_keep]
test_df_filtered = test_df[columns_to_keep]

# Step 3: Retrain Ridge Regression on the filtered train_df
X_train_filtered = train_df_filtered.drop(columns=['date', 'R'])
y_train_filtered = train_df_filtered['R']

ridge_model_filtered = Ridge(alpha=10000)
ridge_model_filtered.fit(X_train_filtered, y_train_filtered)

# Step 4: Predict on the filtered test_df
X_test_filtered = test_df_filtered.drop(columns=['date', 'R'])
y_test_filtered = test_df_filtered['R']
y_pred_filtered = ridge_model_filtered.predict(X_test_filtered)

# Calculate R² score
r2_filtered = r2_score(y_test_filtered, y_pred_filtered)

# Step 5: Plot predictions vs actual R values
plt.figure(figsize=(12, 6))
plt.plot(test_df_filtered['date'], y_test_filtered, label='Actual R', color='blue', linewidth=1.5)
plt.plot(test_df_filtered['date'], y_pred_filtered, label='Predicted R', color='red', linestyle='--', linewidth=1.5)
plt.xlabel('Date')
plt.ylabel('R')
plt.title(f'Ridge Regression Predictions vs Actuals (Top 200 Features, R² = {r2_filtered:.4f})')
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

r2_filtered
```

```python
from sklearn.ensemble import RandomForestRegressor
import optuna
def objective_rf(trial):
    n_estimators = trial.suggest_int("n_estimators", 100, 1000, step=100)
    max_depth = trial.suggest_int("max_depth", 5, 50, step=5)
    min_samples_split = trial.suggest_int("min_samples_split", 2, 10)
    min_samples_leaf = trial.suggest_int("min_samples_leaf", 5, 10)

    rf = RandomForestRegressor(
        n_estimators=n_estimators,
        max_depth=max_depth,
        min_samples_split=min_samples_split,
        min_samples_leaf=min_samples_leaf,
        random_state=42,
        n_jobs=-1,
    )

    rf.fit(X_train_filtered, y_train_filtered)
    y_pred_filtered_rf = rf.predict(X_test_filtered)
    return r2_score(y_test_filtered, y_pred_filtered_rf)
study = optuna.create_study(direction="maximize")
study.optimize(objective_rf, n_trials=20)

print("Best parameters:", study.best_params)
print("Best R-Squared:", study.best_value)
```

```python
def timing_strategy_evaluation_with_drawdown(trained_model, X_test, actual_returns, risk_free_rate=0.02 / 12, threshold=0.01, initial_value=100):
 
    # Predict returns using the trained model
    predicted_returns = trained_model.predict(X_test)

    # Initialize portfolio for timing strategy
    portfolio_values_timing = [initial_value]

    # Timing strategy
    for i in range(len(predicted_returns)):
        if predicted_returns[i] > threshold:  # Risk-On
            portfolio_values_timing.append(portfolio_values_timing[-1] * (1 + actual_returns[i+actual_returns.index.min()]))
        else:  # Risk-Off
            portfolio_values_timing.append(portfolio_values_timing[-1] * (1 + risk_free_rate))

    # Convert portfolio values to pandas Series for analysis
    portfolio_values_timing = pd.Series(portfolio_values_timing)
    # Calculate performance metrics for timing strategy
    cumulative_return_timing = portfolio_values_timing.iloc[-1] / portfolio_values_timing.iloc[0] - 1
    sharpe_ratio_timing = (portfolio_values_timing.pct_change().mean() - risk_free_rate) / portfolio_values_timing.pct_change().std()

    # Calculate maximum drawdown for timing strategy
    rolling_max_timing = portfolio_values_timing.cummax()
    drawdown_timing = (portfolio_values_timing - rolling_max_timing) / rolling_max_timing
    max_drawdown_timing = drawdown_timing.min()

    # Buy-and-hold strategy
    portfolio_values_bh = [initial_value]
    for ret in actual_returns:
        portfolio_values_bh.append(portfolio_values_bh[-1] * (1 + ret))
    portfolio_values_bh = pd.Series(portfolio_values_bh)


    # Plot portfolio evolution
    plt.figure(figsize=(10, 6))
    plt.plot(portfolio_values_timing, label="Timing Strategy", marker='o', linestyle='-')
    plt.plot(portfolio_values_bh, label="Buy-and-Hold Strategy", marker='x', linestyle='--')
    plt.title(f'Portfolio Evolution: Timing Strategy vs Buy-and-Hold for {trained_model}')
    plt.xlabel("Time (Months)")
    plt.ylabel("Portfolio Value")
    plt.legend()
    plt.grid()
    plt.show()

    return portfolio_values_timing, cumulative_return_timing, sharpe_ratio_timing, max_drawdown_timing
```

```python
portfolio_values, cumulative_return, sharpe_ratio, max_drawdown = timing_strategy_evaluation_with_drawdown(
    ridge_model_filtered, X_test_filtered, y_test_filtered
)

# Display results
print("Cumulative Return:", round(cumulative_return * 100, 2), "%")
print("Sharpe Ratio:", round(sharpe_ratio, 2))
print("Maximum Drawdown:", round(max_drawdown * 100, 2), "%")
```
